Task 1 — Derivative and slope intuition 
Take the function 𝑓(𝑥) = 𝑥2 + 2𝑥 +1. Compute the numerical derivative at 3 chosen points, 
then plot the curve and draw the tangent lines at those points. Write 3 short observations on 
how the slope changes and what positive, negative, and near-zero slope mean in ML 
optimization. 
Required items: Python notebook, NumPy, matplotlib.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-10, 10, 100)
f_x = x**2 + 2*x + 1
plt.plot(x, f_x, label='f(x) = x^2 + 2x + 1')
points = [-5, 0, 5]
```

```python
for point in points:
    h = 0.01
    idx = np.argmin(np.abs(x - point))
    x[idx] = point
    f_x[idx] = point**2 + 2*point + 1
    derivative = ((point + h)**2 + 2*(point + h) + 1 - ((point - h)**2 + 2*(point - h) + 1)) / (2 * h)
    tangent_line = derivative * (x - point) + f_x[np.where(x == point)[0][0]]
    plt.plot(x, tangent_line, label=f'Tangent at x={point}')
    plt.legend()
    plt.title('Plot of f(x) and its tangents at chosen points')
    plt.xlabel('x')
    plt.ylabel('f(x)')
    plt.grid()
    plt.show()  
```

Observations:
1. The slope of the tangent line at x=-5 is positive, indicating that the function is increasing at that point. In ML optimization, a positive slope suggests that we are moving away from the minimum, and we need to adjust our parameters in the opposite direction to minimize the loss.
2. The slope of the tangent line at x=0 is zero, indicating that the function has a local minimum at that point. In ML optimization, a near-zero slope suggests that we are close to the optimal parameters, and further adjustments may yield diminishing returns.
3. The slope of the tangent line at x=5 is positive again, indicating that the function is increasing at that point. In ML optimization, a positive slope at this point suggests that we are moving away from the minimum, and we need to adjust our parameters in the opposite direction to minimize the loss.
