Task 3 — Probability basics with sampled data 
Generate two sample sets: one from a uniform distribution and one from a normal 
distribution. Calculate mean, standard deviation, and expectation-style interpretation for 
both. Plot histograms and explain the difference in spread and shape in simple words. 
Required items: Python notebook, NumPy, matplotlib. 

```python
import numpy as np
import matplotlib.pyplot as plt
import IPython.display as display
np.random.seed(42)
num_samples = 1000
uniform_data = np.random.uniform(0, 10, num_samples)
normal_data = np.random.normal(5, 2, num_samples)
```

```python
u_mean = np.mean(uniform_data)
u_std = np.std(uniform_data)
n_mean = np.mean(normal_data)
n_std = np.std(normal_data)
print("--- Statistical Summary (N = 1000) ---")
print(f"Uniform Distribution U(0, 10):")
print(f"  Calculated Mean    : {u_mean:.4f} (Theoretical: 5.0000)")
print(f"  Calculated Std Dev : {u_std:.4f} (Theoretical: 2.8867)")
print(f"Normal Distribution N(5, 2^2):")
print(f"  Calculated Mean    : {n_mean:.4f} (Theoretical: 5.0000)")
print(f"  Calculated Std Dev : {n_std:.4f} (Theoretical: 2.0000)\n")
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5), sharey=False)
```

```python
ax1.hist(uniform_data, bins=20, color='skyblue', edgecolor='black', alpha=0.7, density=True)
ax1.axvline(u_mean, color='red', linestyle='dashed', linewidth=2, label=f'Mean ({u_mean:.2f})')
ax1.set_title('Uniform Distribution $U(0, 10)$', fontsize=12)
ax1.set_xlabel('Value', fontsize=10)
ax1.set_ylabel('Density', fontsize=10)
ax1.grid(True, linestyle=':', alpha=0.6)
ax1.legend()
```

```python
ax2.hist(normal_data, bins=20, color='lightgreen', edgecolor='black', alpha=0.7, density=True)
ax2.axvline(n_mean, color='red', linestyle='dashed', linewidth=2, label=f'Mean ({n_mean:.2f})')
ax2.set_title('Normal Distribution $\mathcal{N}(5, 2^2)$', fontsize=12)
ax2.set_xlabel('Value', fontsize=10)
ax2.set_ylabel('Density', fontsize=10)
ax2.grid(True, linestyle=':', alpha=0.6)
ax2.legend()
```

```python
plt.tight_layout()
display.display(fig)
plt.show()
```

Difference in Shape

Uniform Distribution (Flat/Rectangular): The histogram resembles a flat, rectangular block. Every partition (bin) between $0$ and $10$ contains roughly the same frequency of points ($1000 \text{ samples} / 20 \text{ bins} \approx 50 \text{ samples/bin}$). This flat profile reflects equal probability.

Normal Distribution (Symmetric/Bell-Shaped): The histogram forms a classic symmetric, bell-shaped curve peaking at the mean ($5.0$) and tapering down gradually toward the edges. This visually demonstrates central tendency.

Difference in Spread

Uniform Distribution (Bounded & Dispersed): The uniform data has a wider spread ($s \approx 2.90$) because data points are dispersed uniformly all the way to the boundaries of the interval $[0, 10]$. There is no clustering.

Normal Distribution (Unbounded & Concentrated): The normal distribution has a narrower spread ($s \approx 1.96$). By the Empirical Rule ($68-95-99.7$), approximately $68\%$ of the data lies closely clustered within one standard deviation of the mean ($3.0$ to $7.0$), making the overall distribution much more compact around the center.
